```python
import numpy as np
import sys
import os

t2_path = os.path.abspath('../T2')
sys.path.append(t2_path)

from matstat_facilities import *
from scipy import stats

beta = 0.95
teta = 5
n = 100
u = np.random.uniform(teta, 2 * teta, n)

teta_est_omm = 2/3 * get_ar_mean(u)
teta_est_omp = np.max(u) / 2
teta_est_omp_corr = (2 * n + 2)/(2 * n + 1) * teta_est_omp
```

```python
def get_accur_teta_inter():
    l_border = np.max(u) / (1 + ((1 - beta) / 2)**(-1/n))
    r_border = np.max(u) / (1 + ((1 + beta) / 2)**(-1/n))
    return l_border, r_border
```

```python
def get_asimp_teta_inter():
    norm_left = stats.norm.ppf((1 + beta) / 2)
    norm_right = stats.norm.ppf((1 - beta) / 2)
    l_border = teta_est_omm - norm_left * np.sqrt(4/9 * (get_ar_mean(u*u) - get_ar_mean(u)**2)) / np.sqrt(n)
    r_border = teta_est_omm - norm_right * np.sqrt(4/9 * (get_ar_mean(u*u) - get_ar_mean(u)**2) / np.sqrt(n))
    return l_border, r_border
```

```python
def get_interv_info(interv):
    l = interv[0]
    r = interv[1]
    print('(',l,',', r,')')
    print("Длина интервала =", r-l)
```

```python
print("Объём выборки =", n)
print("Доверительная вероятность =", beta)
print("Истинное значение тета =", teta)
print("ОММ параметра тета =", teta_est_omm)
print("ОМП параметра тета =", teta_est_omp)
print("Исправленная ОМП параметра тета =", teta_est_omp_corr)
print("\nДоверительный интервал для параметра тета:")
print("\nТочный дов. инт. тета")
accur_teta_inter = get_accur_teta_inter()
get_interv_info(accur_teta_inter)
print("\nАсимптотический дов. инт. тета")
asimp_teta_interv = get_asimp_teta_inter()
get_interv_info(asimp_teta_interv)
print("\nНепараметрический бутстрап:")
non_par_bootstr_teta_interv = non_param_bootstr_T5(u, n)
get_interv_info(non_par_bootstr_teta_interv)
print("\nПараметрический бутстрап:")
param_bootstr_teta_interv = param_bootstr_T5(u,n)
get_interv_info(param_bootstr_teta_interv)
```
