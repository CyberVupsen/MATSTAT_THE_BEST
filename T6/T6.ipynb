```python
import numpy as np
import sys
import os

t2_path = os.path.abspath('../T2')
sys.path.append(t2_path)

import matstat_facilities
from matstat_facilities import *
from scipy import stats

beta = 0.95
teta = 5
n = 100
u = np.random.uniform(0, 1, n)
x = (1 - u) ** (1 / (1 - teta))

teta_est = 1 + n / (np.sum(np.log(x)))
```

```python
def get_median_inter():
    lnx = np.log(x)
    sum = np.sum(lnx)
    hi_left = stats.chi2.ppf((1 + beta) / 2, df=2*n)
    hi_right = stats.chi2.ppf((1 - beta) / 2, df=2*n)
    l_border = 2 ** (2 * sum / hi_left)
    r_border = 2 ** (2 * sum / hi_right)
    return l_border, r_border
```

```python
def get_asimp_teta_inter():
    norm_left = stats.norm.ppf((1 + beta) / 2)
    norm_right = stats.norm.ppf((1 - beta) / 2)
    l_border = teta_est - norm_left * (teta_est - 1) / np.sqrt(n)
    r_border = teta_est - norm_right * (teta_est - 1) / np.sqrt(n)
    return l_border, r_border
```

```python
def get_interv_info(interv):
    l = interv[0]
    r = interv[1]
    print('(',l,',', r,')')
    print("Длина интервала =", r-l)
```

```python
print("Объём выборки =", n)
print("Доверительная вероятность =", beta)
print("\nМедиана выборки =", 2**(1/(teta - 1)))
print("Доверительный интервал для медианы:")
print("Точный метод:")
accure_med_interv = get_median_inter()
get_interv_info(accure_med_interv)
print("\nНепараметрический бутстрап:")
non_par_bootstr_med_interv = non_param_bootstr_median(x,n)
get_interv_info(non_par_bootstr_med_interv)
print("\nПараметрический бутстрап:")
param_bootstr_med_interv = param_bootstr_median(x, n)
get_interv_info(param_bootstr_med_interv)
```

```python
print("\nДоверительный интервал для параметра тета:")
print("Истинное значение тета =", teta)
print("Оценка параметра тета =", teta_est)
print("\nАсимптотический метод")
asimp_teta_interv = get_asimp_teta_inter()
get_interv_info(asimp_teta_interv)
print("\nНепараметрический бутстрап:")
non_par_bootstr_teta_interv = non_param_bootstr_teta(x, n)
get_interv_info(non_par_bootstr_teta_interv)
print("\nПараметрический бутстрап:")
param_bootstr_teta_interv = param_bootstr_theta(x,n)
get_interv_info(param_bootstr_teta_interv)
```
